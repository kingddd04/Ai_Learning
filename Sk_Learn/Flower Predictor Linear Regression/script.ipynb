# Linear Regression Example — ML

In this small notebook, we explore a **linear regression model**, one of the simplest and most widely used machine learning algorithms for predicting **continuous** numerical values.

It works by finding the **best‑fitting straight line** through the data that represents the relationship between:
- **Input features (X)** → the independent variables you use to make predictions  
- **Target variable (y)** → the dependent variable you are trying to predict  

## 📈 What is Linear Regression?

Linear regression is a simple yet powerful algorithm used to **predict a continuous value** based on one or more input features.  
It works by finding the **best‑fitting straight line** that describes the relationship between:

- **X** → the input features (independent variables)  
- **y** → the target value you want to predict (dependent variable)  

The line is defined by:
\[
y = m x + b
\]
- **m** → slope (how much y changes when x changes)  
- **b** → intercept (value of y when x = 0)  

**Goal:** Minimize the difference between the model’s predictions and the actual values — often measured using **Mean Squared Error (MSE)**.

**Example:** Predicting house prices from features like size, location, and number of rooms.
📌 **ML Model Creation**
1. Load a dataset and optionally preprocess it (handle missing values, scale data, etc.).
2. Separate the dataset into:
   - **X** → the feature matrix (inputs)
   - **y** → the target vector (outputs)
3. Create a linear regression model.
4. Train the model using your training data so it learns the slope and intercept.
5. Use the model to make predictions and interpret the results.

```python
# let's import sklearn an ai library and pandas a data managment library
from sklearn.linear_model import LogisticRegression
import pandas
```

```python
iris_datas = pandas.read_csv("iris.csv") # Let's load the datas contained into the csv file into a dataframe, an object used by pandas
iris_datas.head() # displaing the beginning of the dataframe
```

```python
"""
In this block of code, we are preparing the dataset that our model will be trained on.
This type of training is called supervised learning, which means we need to provide the model
with both the input features and the corresponding expected outputs (labels).
We create a DataFrame containing only the features of the flower,
and a separate Series containing the target labels.
"""
flower_datas = iris_datas.drop(columns=["variety"])  # Extracting the features by dropping the target column
flower_variety = iris_datas["variety"]               # Extracting the target labels in a new dataframe
print(flower_datas.head())
print(flower_variety.head())
```

```python
#ML Model creation
model = LogisticRegression(verbose=1, max_iter=200)
```

```python
# The model "learns" here by finding patterns that connect the input features (flower measurements)
# to the correct target labels (flower species). During this training step, it adjusts its internal
# parameters (weights and intercept) so that it can make accurate predictions on new, unseen flowers.
model.fit(flower_datas.values, flower_variety)
```

```python
from sklearn.metrics import accuracy_score

# Getting a prediction
prediction = model.predict([[4.6, 3.5, 1.5, 0.2]])
print(" A flower with sepal.length sepal.width petal.length petal.width variety is probably a : ", prediction)
```
