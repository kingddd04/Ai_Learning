### What is a LLM?

LLM are deep neural networks that are able to process text. In their youth they where based on Recurrent Neural Networks, in 2018 where introduced transformers with their revolutionary inpact, started a new era for text processing tasks.
They are the evolution of Nlp (Natural Language Processing Models) very good in a simple task, LLM are capable of manage a lot of testual tasks and not only, like the newest multimodal models, that can process audios and images.

- **Definition:**  
  An LLM is a machine learning model that learns statistical patterns of language, enabling tasks such as text completion, translation, summarization, and question answering.

- **Examples of LLMs:**  
  - GPT (Generative Pre-trained Transformer)  
  - BERT (Bidirectional Encoder Representations from Transformers)  
  - LLaMA (Large Language Model Meta AI)

> 💡 **Note:** LLMs are not just about memorizing text. They learn **representations of meaning**, allowing them to reason, infer, and generate coherent responses across contexts.

### 🧪 Pre-training and Fine-tuning in LLMs

Large Language Models (LLMs) are built through a **two-stage training process**:  
1. **Pre-training**  
2. **Fine-tuning**  

**📘 1. Pre-training**
Pre-training is the **foundation stage** where the model learns general language patterns.

- **Goal:** Teach the model statistical properties of language (grammar, semantics, context).  
- **Method:** Train on massive, diverse text corpora (books, articles, websites).  
- **Task:** Usually a **self-supervised objective**, such as predicting the next word (causal language modeling) or filling in missing words (masked language modeling).  

- **Outcome:**  
  - The model acquires broad knowledge of language (or whatever you fed to it).  
  - It becomes capable of generating coherent text (or whatever you fed to it) and understanding context  of whatever you fed to it.

**⚙️ 2. Fine-tuning**
Fine-tuning is the **specialization stage** where the model adapts to specific tasks or domains.

- **Goal:** Align the model with desired applications (e.g., medical text analysis, customer support).  
- **Method:** Train on smaller, task-specific datasets.  
- **Types of fine-tuning:**  
  - **Supervised fine-tuning:** Using labeled data (e.g., sentiment classification).  
  - **Instruction tuning:** Training the model to follow human instructions.  
  - **Reinforcement Learning from Human Feedback (RLHF):** Aligning outputs with human preferences.  

**So to summarize:**
**Pre-training:** Model learns general English from billions of sentences.    
**Fine-tuning:** Model is further trained on legal documents to become a legal assistant.
