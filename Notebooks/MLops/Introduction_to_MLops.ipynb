### What is Mlops? 

When we refer to mlops we talk about a discipline that links Machine learning and devOPS. Usually a ml project is not only the statistical model but is used in an application, that runs on a server, needs some datas to train and must fulfill it's duties in a certin context.
So is needed a mature programming approach made of classes, packages, versioning and automation. This is necessary because ai and ml systems are dynamic and must resist the trial of time and work in different enviorments.

### How are Batch systems classified?

An Ai system may be classified by:
- how datas arrive
- how predictions are done
- What operational constraints must be respected(cost, latency, performance, precision)

**1. Batch Systems**

A batch system elaborates datas in sets determined by a certain time, or complete datasets ciclically. Predictions are generated by jobs(finite executons) and saved in a desired storage to be used.

- Latency: The answer is ot in real time.
- Inference: Runs as a job, that reads some features and does a prediction 

**2. Streaming systems**

Is a system that operates continously,  data arrives as a stream of datas and the system must react fast to queries; In this cases often a Hybrid architecture is followed, Offline features+ online features. A real life example may be fraud detection systems.

- Latency : minimal, barely tollerated, depending on context.
- Error Managment and Events, datas streams often may include errors, duplicates and issues that need to be managed.
- Continous stress: the project must be resilient and easily monitored.

**3. Gen-Ai Systems/LLM-based systems**

This architecture orbit around a generator model, that often is completed by tools that assist it in various fields like fact checking and user documents retrival. 

- Cost and performance: These models must balance computational efficiency with high‑quality outputs.
- Context and external knowledge: They rely on extended context windows and external sources(tools like rag) to provide accurate, grounded responses.
- Security and privacy : Control of output for not allowed generations, and respect of the user privacy

### The Paradigm : Feature, training and inference pipeline:

The FTI paradigm provides a clear model for designing and governing the entire lifecycle of an AI/ML-based predictive system.
The core value of this approach lies in its ability to reduce the apparent complexity of machine learning systems. 
Such systems are often perceived as a fragmented collection of components, tools, and processes. The FTI paradigm simplifies this view by organizing the system around three fundamental pipelines, each with well-defined responsibilities, inputs and outputs, and operational constraints.

According to this perspective, any AI/ML system can be understood as the interaction of three main subsystems.

- The Feature Pipeline is responsible for computing and making features available from raw data. It defines how data is ingested, transformed, validated, and exposed so that features are consistent and reusable across the system.

- The Training Pipeline uses historical features and labels to train, evaluate, and select a model. This pipeline focuses on experimentation, model comparison, and reproducibility, ensuring that the chosen model meets the desired performance criteria.

- The Inference Pipeline applies the trained model to new data in order to produce predictions. This pipeline typically operates under strict requirements for reliability

### The Training Pipeline

Is responsible for producing a deployable and reproducible model artifact starting from historical features and labels. Its role extends beyond model fitting: it enforces correctness, temporal discipline, and methodological rigor across the entire training lifecycle.  
Its tasks are:

- Dataset construction: Joining historical features with labels, applying sampling strategies, handling imbalance, and filtering invalid or noisy samples  
- Temporal correctness: Ensuring point‑in‑time validity to prevent data leakage  
- Data splitting: Creating training, validation, and test splits, often time‑based  
- Model training and selection: Running controlled training loops, hyperparameter tuning, and evaluating candidate models  
- Packaging and registration: Serializing the model, generating metadata, defining schemas, and tracking dependencies

These may be the typical components, classes that are included in this pipeline:

- Dataset Builder: Constructs the training dataset by merging features and labels while enforcing temporal correctness  
- Split Manager: Defines and applies time‑based or cross‑validation splits  
- Trainer: Executes training loops, hyperparameter searches, and metric evaluations  
- Evaluator: Computes performance metrics, robustness checks, and acceptance criteria  
- Model Registry: Stores the selected model along with metadata, versioning, and lineage information

### Inference Pipeline

Is responsible for applying a trained and registered model to new incoming data in order to produce predictions under strict operational constraints. It ensures consistency with the training process while meeting latency, availability, and reliability requirements.  
Its tasks are:

- Feature retrieval: Loading or computing features using the same definitions as in training  
- Model loading: Selecting and loading the correct model version for prediction  
- Prediction execution: Running deterministic and efficient inference logic  
- Post‑processing: Applying thresholding, ranking, calibration, or business rules  
- Monitoring: Emitting signals for latency, errors, drift, and model degradation

These may be the typical components, classes that are included in this pipeline:

- Online Feature Loader: Retrieves low‑latency features from an online store or cache  
- Model Loader: Loads the appropriate model version and its associated metadata  
- Predictor: Executes the model’s forward pass and produces raw outputs  
- Post‑Processor: Converts raw outputs into final predictions for downstream consumers  
- Monitoring Layer: Tracks operational metrics, drift indicators, and prediction behavior over time
