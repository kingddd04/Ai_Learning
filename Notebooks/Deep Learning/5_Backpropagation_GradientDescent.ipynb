### Backpropagation:

Backpropagation is the fundamental algorithm for training neural networks. It computes how much each weight and bias contributes to the final error by propagating error signals backward through the network. Using the chain rule of calculus, it efficiently calculates gradients layer by layer, enabling optimization methods like gradient descent to update parameters and reduce loss. Despite being introduced over 40 years ago, backpropagation remains the backbone of modern deep learning.

Without backpropagation, computing gradients would require recalculating the loss for each parameter separately—impractical for millions of weights.  
Backpropagation itself doesn’t update weights it only computes gradients that will be used by the optimization algoritm.

### Gradient Descent:

Gradient descent is a weight optimization technique that takes the gradient of each weight (the one that represent how much each weight is responsible for the error) and uses it to optimize weights afer being multiplied by a constant.

**formula** : weight = weight- constant_learning_rate * gradient
