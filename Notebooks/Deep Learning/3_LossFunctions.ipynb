### 🔍The Loss Function in Neural Networks

In neural networks, a **loss function** is a mathematical tool that measures how far off a model's predictions are from the actual target values. It acts as a compass during training for the backpropagation guiding the model toward better performance by minimizing prediction errors.

Loss functions are crucial because they:

- 📉 Quantify prediction error
- 🔁 Drive weight updates via backpropagation
- 🎯 Influence learning behavior and convergence

**🧠 Why Do We Need Loss Functions?**

Imagine teaching a student to solve math problems. After each attempt, you give feedback on how close they were to the correct answer. The loss function plays this role for neural networks—it tells the model *how wrong* it was and helps it improve.

**🧪 Types of Loss Functions**

Loss functions vary depending on the task:

- **Regression** → Predicting continuous values (e.g., house prices)
- **Classification** → Predicting discrete labels (e.g., cat vs dog)
- **Ranking / Metric Learning** → Comparing similarity or order
- **Segmentation / Spatial Tasks** → Evaluating overlap in images
- **Custom / Composite** → Tailored to specific domains or multi-task setups

Each type has multiple variants, chosen based on the problem’s nature and data characteristics.
Let's see the most common ones 

### MSE LOSS

**🔹 Definition**  
The **Mean Squared Error (MSE)** is a common loss function used in **regression tasks**.  
It measures the average squared difference between the predicted values and the actual target values.

formula = sum

- Think of MSE as measuring how far your predictions are from the truth.  
- Each error is **squared**, so:
  - Negative and positive errors don’t cancel out.  
  - Larger errors are penalized more heavily.  

👉 This makes MSE very sensitive to **outliers**.

```python
def mean_squared_error(y_true, y_pred):
    n = len(y_true)
    total_error = 0
    for i in range(n):
        total_error += (y_true[i] - y_pred[i]) ** 2
    return total_error / n

# Example
y_true = [2, 3, 5]
y_pred = [2.5, 2.0, 5.5]

print("Mean Squared Error:", mean_squared_error(y_true, y_pred))
```
