### Overfitting Reduction Techniques

Neural Networks in general have a issue they tend to learn, well, maybe too well specific patterns, for this reason they perform well on specific paths but they kinda struggle with generalization, for example an ai being trained to reconize yellow cats over a set of animals when seeing a red cat it may not even reconize it as a cat but maybe as a parrot.

To address this we have different methods :

-Regularization(l1, l2)  
-Dropout  
-Data Augmentation  
-Early Stopping   

### Dropout: What it is and How it Works

Dropout is a regularization technique proposed by Geoffrey Hinton and others in 2014 (Dropout: A Simple Way to Prevent Neural Networks from Overfitting).

During training, each neuron has a probability 𝑝
 of being “switched off,” meaning it does not participate in the computation. This prevents the network from relying too heavily on specific neurons and encourages it to learn more robust representations.

During training: a portion of the neurons is randomly deactivated.

During prediction: all neurons are active, but their values are scaled to maintain the same statistical intensity.

Advantages:

1. Reduces co-adaptation between neurons (neighboring neurons learning overly similar  features
2. Forces the network to learn general patterns.
3. Forces the network to learn more general patterns.
Improves generalization ability, even with small datasets.

### Data Augmentation

Data augmentation is a technique that aims to edit randomly the datas, in a way that the model may learn to generalize or at least support a certain amount of variation.
For example in an image context images may be zoomed in or out, cropped, color filtered, flipped, by random amounts in many random ways.
