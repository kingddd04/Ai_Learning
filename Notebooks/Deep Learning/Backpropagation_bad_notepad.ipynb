# Backpropagation

```python
def relu(x):
    return np.maximum(0, x)
```

```python
X = np.array([
    [0.1, 0.5, 1.2],
    [0.3, 1.8, -0.3],
    [-0.4, 0.9, 1.4],
])

y_reg = np.array([[1.689],
                  [1.088],
                [0.886]])

y_bin = np.array([[0],
                  [1],
                  [0]])

y_mul = np.array([[1, 0, 0],
                 [0, 1, 0],
                 [1, 0, 0]])

B = X.shape[0]
n = 3
```

```python
# Primo hidden layer = 4 neuroni

W1 = np.random.randn(4, 3)
b1 = np.random.randn(4)
```

```python
Z1 = X @ W1.T + b1
```

```python
A1 = relu(Z1)
```

```python
# Secondo hidden layer

W2 = np.random.randn(4, 4)
b2 = np.random.randn(4)
```

```python
Z2 = A1 @ W2.T + b2
A2 = relu(Z2)
```

# Output Layer

## Regressione

```python
W3 = np.random.randn(1, 4)
b3 = np.random.randn(1)
```

```python
Y_hat = A2 @ W3.T + b3
```

```python
print(Y_hat)
```

# Backward pass

```python
def mse_loss(y_hat, y):
    return np.mean((y_hat - y) ** 2)
```

```python
loss = mse_loss(Y_hat, y_reg)
```

```python
print(loss)
```

```python
dZ3 = (2.0 / B) * (Y_hat - y_reg)
```

```python
print(dZ3)
```

## Classificazione binaria

##  output layer

```python
def sigmoid(x):
    return 1.0 / (1.0 + np.exp(-x))
```

```python
def bce_loss_from_logits(z, y):
    a = np.maximum(z, 0)
    return np.mean(a - z * y + np.log1p(np.exp(-np.abs(z))))
```

```python
Z3 = A2 @ W3.T + b3
```

```python
p = sigmoid(Z3)
```

```python
print(p)
```

```python
loss = bce_loss_from_logits(Z3, y_bin)
```

```python
print(loss)
```

```python
dZ3 = (p - y_bin) / B
```

```python
print(dZ3)
```

# Classificazione Multiclasse

# output layer

```python
K = 3
```

```python
def softmax_batch(z):
    z = z - z.max(axis=1, keepdims=True)
    e = np.exp(z)
    return e / e.sum(axis=1, keepdims=True)
```

```python
W3 = np.random.randn(K, 4)
b3 = np.random.randn(K)
```

```python
Z3 = A2 @ W3.T + b3
```

```python
P = softmax_batch(Z3)
```

```python
print(P)
```

```python
def cross_entropy_from_logits(z, y1h):
    z2 = z - np.max(z, axis=1, keepdims=True)
    lse = np.log(np.sum(np.exp(z2), axis=1, keepdims=True))
    log_p = z2 - lse
    return -np.mean(np.sum(y1h * log_p, axis=1, keepdims=True))
```

```python
loss = cross_entropy_from_logits(Z3, y_mul)
```

```python
print(loss)
```

```python
dZ3 = (P - y_mul) / B
```

```python
print(dZ3)
```

# Esempio Completo su tutto il dataset (Batch Gradient Descent)

```python
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_regression
```

### Regressione

```python
X, y = make_regression(n_samples=300, n_features=3, noise=15.0, random_state=42) # dataset sintetico
```

```python
print(f"Dimensione iniziale di y:", y.shape)
```

```python
y = y.reshape(-1,  1) # reshape di y
```

```python
print(f"Dimensione di y dopo il reshape:", y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
B, d = X_train.shape # B = dimensione del batch (tutto il dataset), # d = numero di features

print(f"Batch dimension: {B} | n features = {d}")
```

### 1. Inizializzazione

```python
# 2 hidden layer, da 4 neuroni ciascuno, attivazione ReLU

n_neurons_hidden1 = 20
n_neurons_hidden2 = 20
```

```python
# Notazione per la matrice dei pesi del generico layer: W.shape = (n_out, n_in)

W1 = np.random.randn(n_neurons_hidden1, d) * 0.1 # Matrice dei pesi dell' hidden layer 1
b1 = np.zeros((1, n_neurons_hidden1)) # array dei bias dell' hidden layer 1
```

```python
print(f"Dimensioni della matrice dei pesi del layer 1 (Fully connected Layer): {W1.shape}")
print(f"dimensioni vettore bias: {b1.shape}")
```

```python
W2 = np.random.randn(n_neurons_hidden2, n_neurons_hidden1) * 0.1 # Matrice dei pesi dell' hidden layer 2
b2 = np.zeros((1, n_neurons_hidden2)) # array dei bias dell' hidden layer 2
```

```python
print(f"Dimensioni della matrice dei pesi del layer 2 (Fully connected Layer): {W2.shape}")
print(f"dimensioni vettore bias: {b2.shape}")
```

```python
W3 = np.random.randn(1, n_neurons_hidden2) * 0.1 # Matrice dei pesi dell' hidden layer 3 (output layer)
b3 = np.zeros((1, 1)) # array dei bias dell' hidden layer 3 (output layer)
```

```python
print(f"Dimensioni della matrice dei pesi del layer 3 (Fully connected Layer, Output): {W3.shape}")
print(f"dimensioni vettore bias: {b3.shape}")
```

### 2. Forward pass

 - Utilizzando la notazione dimensionale (n_out, n_in) per le matrici dei pesi dei layer, è necessario utilizzare la loro trasposta nel forward pass
  - è possibile inoltre notare come la dimensione delle matrici dei pesi e bias di ciascun layer (e quindi il numero dei parametri della rete), siano unicamente influenzate dall'architettura della rete neurale, mentre la dimensione del batch influenza le dimensioni dell'output dei vari livelli (Le matrici Z e A)

```python
Z1 = X_train @ W1.T + b1 # Pre-attivazioni del primo hidden layer
A1 = relu(Z1) # Attivazioni (output) del primo hidden layer
```

```python
print(f"Dimensioni matrice delle pre attivazioni (Z1) layer 1 (Fully connected Layer): {Z1.shape}")
print(f"Dimensioni matrice delle attivazioni (A1) layer 1 (Fully connected Layer): {A1.shape}")
```

```python
Z2 = A1 @ W2.T + b2
A2 = relu(Z2)
```

```python
print(f"Dimensioni matrice delle pre attivazioni (Z2) layer 2 (Fully connected Layer): {Z2.shape}")
print(f"Dimensioni matrice delle attivazioni (A2) layer 2 (Fully connected Layer): {A2.shape}")
```

```python
Z3 = A2 @ W3.T + b3
```

```python
print(Z3)
```

```python
y_hat = Z3 # output della rete, regressione.
```

```python
print(f"Dimensioni matrice output della rete neurale (Z3) contenente le predizioni finali del modello: {Z3.shape}")
```

```python
print(f"Predizioni del modello (uguale a Z3 nella regressione):")
print(y_hat.shape)
```

### 3. Calcolo della loss
- nell'esempio si lavora su un problema di regressione, pertanto utilizzeremo il MSE come funzione di perdita da minimizzzare

```python
loss = mse_loss(y_hat, y_train) # loss MSE (Mean Squared Error) in regressione
```

```python
print(f"Loss iniziale: {loss}")
```

### 4. Backward pass

 - Utilizzando la notazione dimensionale (n_out, n_in) per le matrici dei pesi dei layer, non è necessario utilizzare la loro trasposta nel backward pass

```python
dZ3 = (2.0 / B) * (y_hat - y_train) # errore "locale" del layer 3 (output layer)
```

```python
dW3 = dZ3.T @ A2 # gradiente dei pesi del layer 3
```

```python
db3 = dZ3.sum(axis=0, keepdims=True) # gradiente dei bias del layer 3
```

```python
dA2 = dZ3 @ W3
```

### Output layer

### Secondo hidden layer

```python
dZ2 = dA2 * relu_deriv(Z2) # errore "locale" del layer 2 (secondo hidden layer)
```

```python
dW2 = dZ2.T @ A1 # gradiente dei pesi del layer 2
```

```python
db2 = dZ2.sum(axis=0, keepdims=True)
```

```python
dA1 = dZ2 @ W2
```

### Primo hidden layer

```python
dZ1 = dA1 * relu_deriv(Z1)  # errore "locale" del layer 1 (primo hidden layer)
```

```python
dW1 = dZ1.T @ X_train
```

```python
db1 = dZ1.sum(axis=0, keepdims=True)
```

# Aggiornamento dei pesi

```python
eta = 0.1
for W, dW, b, db in [(W1, dW1, b1, db1), (W2, dW2, b2, db2), (W3, dW3, b3, db3)]:
    W -= eta * dW
    b -= eta * db
```

# Secondo forward pass e calcolo della loss con pesi aggiornati

```python
Z1 = X_train @ W1.T + b1 # Pre-attivazioni del primo hidden layer
A1 = relu(Z1) # Attivazioni (output) del primo hidden layer

Z2 = A1 @ W2.T + b2
A2 = relu(Z2)

Z3 = A2 @ W3.T + b3

y_hat = Z3 # output della rete, regressione.
```

```python
loss = mse_loss(y_hat, y_train) # loss MSE (Mean Squared Error) in regressione
```

```python
print(f"Loss dopo aggiornamento: {loss}")
```

# Mini-batch gradient descent

```python
n_neurons_hidden1 = 20
n_neurons_hidden2 = 20
```

```python
# Notazione per la matrice dei pesi del generico layer: W.shape = (n_out, n_in)

W1 = np.random.randn(n_neurons_hidden1, d) * 0.1 # Matrice dei pesi dell' hidden layer 1
b1 = np.zeros((1, n_neurons_hidden1)) # array dei bias dell' hidden layer 1

W2 = np.random.randn(n_neurons_hidden2, n_neurons_hidden1) * 0.1 # Matrice dei pesi dell' hidden layer 2
b2 = np.zeros((1, n_neurons_hidden2)) # array dei bias dell' hidden layer 2

W3 = np.random.randn(1, n_neurons_hidden2) * 0.1 # Matrice dei pesi dell' hidden layer 3 (output layer)
b3 = np.zeros((1, 1)) # array dei bias dell' hidden layer 3 (output layer)
```

```python
print(f"Totale parametri da ottimizzare durante il training:")
print("----------------------------------------------------")

total = 0

for num_layer, layer_params in enumerate([(W1, b1), (W2, b2), (W3, b3)]):
    n_single_weights = layer_params[0].size
    n_biases = layer_params[1].size
    total_layer_params = n_single_weights + n_biases
    total += total_layer_params
    print(f"Layer {num_layer+1} | weights: {n_single_weights} | biases: {n_biases} | tot: {total_layer_params}")
    print("----------------------------------------------------")

print(f"Totale parametri del modello: {total}")
```

```python
def iterate_minibatches(xt, yt, size):
    idx = np.arange(xt.shape[0])
    for s in range(0,  len(idx), size):
        e  = s + size
        bs = idx[s:e]
        yield xt[bs], yt[bs]
        
x = True        
lr, batch_size, epochs = 0.0001, 32, 300
for epoch in range(epochs):
    losses = []
    for Xb, Yb in iterate_minibatches(X_train, y_train, batch_size):
        B = Xb.shape[0]

        # Forward Pass
        Z1 = Xb @ W1.T + b1
        A1 = relu(Z1)
        Z2 = A1 @ W2.T + b2
        A2 = relu(Z2)
        Z3 = A2 @ W3.T + b3
        y_hat = Z3

        # calcoliamo la loss

        loss = mse_loss(y_hat, Yb)
        losses.append(loss)

        # Backward pass
        d3 = (2.0 / B) * (y_hat - Yb) # errore "locale" del layer 3 (output layer)
        dW3 = d3.T @ A2
        db3 = d3.sum(axis=0, keepdims=True)

        dA2 = d3 @ W3
        if x:
            print(dA2)
            x = False
        d2 = dA2 * relu_deriv(Z2)
        dW2 = d2.T @ A1
        db2 = d2.sum(axis=0,  keepdims=True)

        dA1 = d2 @ W2
        d1 = dA1 * relu_deriv(Z1)
        dW1 = d1.T @ Xb
        db1 = d1.sum(axis=0, keepdims=True)

        # update dei pesi
        for W, dW, b, db in [(W1, dW1, b1, db1), (W2, dW2, b2, db2), (W3, dW3, b3, db3)]:
            W -= lr * dW
            b -= lr * db
```
