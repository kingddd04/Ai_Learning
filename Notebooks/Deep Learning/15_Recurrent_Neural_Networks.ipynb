### 🌀 Recurrent Neural Networks (RNNs)

Recurrent Neural Networks are a class of neural architectures designed to handle sequential data (e.g., text, time series, speech). Unlike standard feed‑forward networks, RNNs introduce recurrence: they maintain a hidden state that evolves over time, allowing the network to "remember" information from previous steps.
The information flows backward using a process called feedback.

**Key Parts :**
1. **Recurrent_Unit**
  This is the basic building block of an RNN.
  Unlike a standard neuron, it maintains a Hidden State.
  The hidden state acts as a form of short-term memory, storing a compressed trace of the inputs seen so far.
  At each time step, the hidden state is updated, allowing the network to “remember” past information while processing new inputs.

2. **Recurrent_Unfolding :**
  RNNs are often represented as a loop, since the same unit is applied repeatedly across time.

3. Tanh function : RNNs often use the tanh activation because it keeps hidden states stable by bounding values between
   -1 and 1, preventing uncontrolled growth or decay of signals across  time.

**✨ Key Ideas :**

- Sequentiality: RNNs process inputs one step at a time, updating their hidden state at each step.

- Recurrence: Information flows not only forward (input → output) but also backward in the sense that past hidden states influence future computations.

- Memory: The hidden state acts as a short‑term memory, enabling the network to capture temporal dependencies. However, vanilla RNNs struggle with long‑term dependencies due to issues like vanishing/exploding gradients.

- Window: In practice, RNNs operate within an effective context window — the portion of recent sequence steps that meaningfully influence the current prediction. This window is limited by training techniques such as truncated backpropagation through time (BPTT) and by the network’s ability to retain gradients. As a result, vanilla RNNs usually capture short‑range dependencies (a few steps back), while advanced variants like LSTMs and GRUs extend this window to handle longer contexts.

### 🔒 Long Short-Term Memory (LSTM) :

Long Short-Term Memory networks are an advanced variant of recurrent neural networks designed to overcome the limitations of vanilla RNNs. While RNNs struggle with long-term dependencies due to vanishing or exploding gradients, LSTMs introduce a more sophisticated memory system that allows them to retain information across longer sequences.

At the core of an LSTM is the **cell state**, a dedicated memory pathway that runs through the network with minimal modifications. Information flow into and out of this cell state is carefully regulated by **gates** — specialized neural units that decide what to keep, what to update, and what to output.

**✨ Key Ideas :**

- **Cell State (Memory Line):** Acts as a conveyor belt carrying information across time steps. It enables long-term memory retention by allowing gradients to flow more smoothly during training.

- **Forget Gate:** Determines which parts of the previous cell state should be discarded. This prevents irrelevant or outdated information from cluttering memory.

- **Input Gate:** Decides which new information should be added to the cell state. It combines the current input with the previous hidden state to update memory.

- **Output Gate:** Controls what information from the cell state is exposed as the hidden state at each time step, influencing predictions.

- **Gradient Stability:** By combining gating mechanisms with residual-like pathways, LSTMs mitigate vanishing gradients, making them effective for tasks requiring long-range context.

- **Extended Context Window:** Unlike vanilla RNNs, LSTMs can capture dependencies spanning dozens or even hundreds of steps, making them suitable for language modeling, speech recognition, and time series forecasting.

### ⚠️ Problems with LSTMs in Deep Learning

Long Short-Term Memory (LSTM) networks were designed to overcome the vanishing gradient problem in vanilla RNNs, but they still face several practical challenges when applied to modern deep learning tasks.

**✨ Key Issues**

- **Computational Complexity:**  
  LSTMs are heavier than vanilla RNNs because each unit has multiple gates (forget, input, output).  
  → More parameters lead to slower training and higher memory usage, however are liter than transformers

- **Difficulty with Very Long Sequences:**  
  While LSTMs extend the effective memory window compared to RNNs, they still struggle with extremely long dependencies (hundreds or thousands of steps).  
  → Transformers outperform LSTMs here because attention directly connects distant tokens.

- **Training Instability:**  
  Despite mitigating vanishing gradients, LSTMs can still face exploding gradients if not carefully regularized.  
  → Requires gradient clipping, careful initialization, and tuned learning rates.

- **Interpretability:**  
  The gating mechanisms are mathematically elegant but hard to interpret.  
  → Understanding *why* an LSTM remembers or forgets certain information is non-trivial compared to attention-based models.

- **Parallelization Limits:**  
  LSTMs process sequences step by step (sequential recurrence).  
  → This makes them harder to parallelize on GPUs compared to Transformers, which process entire sequences simultaneously.

- **Overfitting:**  
  With many parameters, LSTMs can easily overfit small datasets.  
  → Requires dropout, regularization, or careful architecture design.

**📊 Practical Note**
LSTMs are particularly effective when:
- Modeling long sequences of text (e.g., machine translation, language modeling).  
- Predicting time series with seasonal or delayed dependencies.  
- Processing audio or speech signals where context spans many frames.
