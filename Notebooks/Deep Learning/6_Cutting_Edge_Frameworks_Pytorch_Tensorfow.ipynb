### Deep Learning Frameworks

As we have seen implementing neural netwoks manually is very hard, for this reason is a smart move using libraries developed by someone with a two-three degrees, probably , to ease the creation of our deep learning models. Nowdays the two best and most used libraries are pytorch and tensorflow.

**TensorFlow**: Backed by Google, it’s highly admired for production deployment, More abstract, less code, great for production pipelines.
**PyTorch**: Backed by Meta, it’s the favorite in research labs due to its dynamic, Pythonic style and ease of prototyping. More explicit, Python‑like, great for learning and debugging.

### Installing

Installing these libraries is very easy. For both of them, it is recommended to create a **virtual environment** to avoid dependency conflicts. Here we are doing it on Windows 11

1. **Create a virtual environment**  

   ```python -m venv dl-env```   This command creates an enviorment called dl-env


2. **Activate the enviorment**

    ```.\dl-env\Scripts\activate``` This command activates the enviorment  

    use ```Set-ExecutionPolicy -Scope Process -ExecutionPolicy Bypass``` if windows won't allow you to run scripts  


3. **Install pytorch or tensorflow with their respective command**

   ```pip install torch torchvision torchaudio ``` This command installs pytorch the cpu version, there is also the gpu version
   ```pip install tensorfow ``` This command installs pytorch 
   

4. **Run them in a Jupyter notebooks**

   ``` pip install jupyter ipykernel ``` istall jupyter in your current virtual enviorment

   ```python -m ipykernel install --user --name=_name --display-name "Python (__name)"``` this command allows you to install a custom kernel  that can run with jupyter can be accessed in various ways depending on your ide

Remember that after those commands you have to swich your kernel to the new one you just created
