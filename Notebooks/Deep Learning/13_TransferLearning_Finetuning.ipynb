### Transfer Learning 

Is a deep learning pratice that involves the usage of a pre-trained model, often very skilled in a task, to build another model.

**Workflow:**

- Get the full model weights
- ✅ You freeze the pre-trained layers (e.g., a ResNet or BERT encoder).
- ✅ You add new layers (e.g., a classification head or regression output).
- ✅ You train only the new layers on your task-specific data.
- 🧠 The pre-trained model acts as a fixed feature extractor.

**Example:** Use a pre-trained CNN on ImageNet, chop off the final classification layer, and add a new dense layer for classifying medical images.
Usually this strategy is used when we have not a lot of datas to work with.

### Finetuning

Fine-tuning is a technique that involves **partially retraining a model that has already been pre-trained** on a large dataset (e.g., ImageNet), by **unfreezing some layers** to adapt their weights to a new task or domain.

Unlike simple feature extraction with transfer learning, fine-tuning does **not just reuse features** — it allows the model to **update part of the base network** in a controlled way, aiming for better performance on the new dataset.

This technique is especially useful when:
- The new dataset is **similar** to the one the model was originally trained on.
- You have a **sufficient amount of data** to retrain without risking overfitting.

**Fine-Tuning Workflow**

1. Load a pre-trained model with known weights (e.g., ImageNet).
2. Temporarily freeze all the base layers (the backbone).
3. Unfreeze only some layers (e.g., the deeper convolutional blocks) or the entire model.
4. Train the network with a very small learning rate, to preserve the acquired knowledge while adapting the features to the new classes or domain.

This approach allows you to reuse the learned representations, gradually adapting them to the new problem, with the goal of improving overall accuracy on the new domain.
