# 🫀 Heart Disease Dataset - Feature Description

This dataset contains patient records with clinical attributes used to predict the presence of heart disease.  
Each row corresponds to one patient.


## 📋 Feature Explanations

- **patientid**  
  Unique identifier for each patient (not used for prediction).

- **age**  
  Age of the patient in years.

- **gender**  
  Biological sex of the patient:  
  - `1` = Male  
  - `0` = Female  

- **chestpain** (Chest Pain Type)  
  Type of chest pain experienced:  
  - `0` = Typical angina  
  - `1` = Atypical angina  
  - `2` = Non-anginal pain  
  - `3` = Asymptomatic  

- **restingBP** (Resting Blood Pressure)  
  Resting blood pressure in mm Hg (measured on admission to the hospital).

- **serumcholestrol**  
  Serum cholesterol level in mg/dl.  
  - `0` may indicate missing or unrecorded values.

- **fastingbloodsugar**  
  Fasting blood sugar > 120 mg/dl:  
  - `1` = True  
  - `0` = False  

- **restingrelectro** (Resting Electrocardiographic Results)  
  Results of resting ECG:  
  - `0` = Normal  
  - `1` = ST-T wave abnormality  
  - `2` = Left ventricular hypertrophy  

- **maxheartrate**  
  Maximum heart rate achieved during exercise.

- **exerciseangia** (Exercise-Induced Angina)  
  - `1` = Yes  
  - `0` = No  

- **oldpeak**  
  ST depression induced by exercise relative to rest (numeric, float).  
  Indicates severity of exercise-induced ischemia.

- **slope** (Slope of the Peak Exercise ST Segment)  
  - `0` = Upsloping  
  - `1` = Flat  
  - `2` = Downsloping  

- **noofmajorvessels**  
  Number of major vessels (0–3) colored by fluoroscopy.  
  Higher values indicate more blocked vessels.

- **target**  
  Presence of heart disease:  
  - `1` = Heart disease present  
  - `0` = No heart disease
