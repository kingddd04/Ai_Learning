# Statistical analisis on Lottery extractions(Superenalotto)


### Load Dataset

```python
import pandas as pd
import numpy as np
import scipy.stats as stats

df = pd.read_csv("superenalotto_oct_2025.csv", sep=r"\s+", engine="python")

df_list = df.values.tolist()

print("Estraction numbers : ", len(df))
```

# Theory : each new estraction rarely has numbers contained in the previous 

We will evaluate this idea

```python
rows_without_rep = 0
rows_without_rep_counter = 0
single_consecutive_appearence_counter = 0
double_consecutive_appearence_counter = 0
triple_consecutive_appearence_counter = 0
old_numbers = df_list[0]
rows_without_rep_archive = []

for df_row in df_list[1:]:
    numbers = df_row[1:-2]
    n_consecutives = len(set(old_numbers) & set(numbers))
    if n_consecutives == 0:
        rows_without_rep+=1
        rows_without_rep_counter+=1
    if n_consecutives == 1:
        single_consecutive_appearence_counter+=1
        rows_without_rep_archive.append(rows_without_rep_counter)
        rows_without_rep_counter = 0
    if n_consecutives == 2:
        double_consecutive_appearence_counter+=1
        rows_without_rep_archive.append(rows_without_rep_counter)
        rows_without_rep_counter = 0
    if n_consecutives == 3:
        triple_consecutive_appearence_counter += 1
        rows_without_rep_archive.append(rows_without_rep_counter)
        rows_without_rep_counter = 0
    old_numbers = numbers
rows_without_rep_archive.append(rows_without_rep_counter)



print("Single-number repetitions:", single_consecutive_appearence_counter)
print("Double-number repetitions", double_consecutive_appearence_counter)
print("Triple-number repetitions", triple_consecutive_appearence_counter)
print("No Repetitions ", rows_without_rep)
print("Max rows without repetitions ", max(rows_without_rep_archive))
print("Average rows without repetitions ", round(np.mean(rows_without_rep_archive), 2))
```

# Theory n 2 : Each number might reappear with a "cooldown" and likely not in the next n extractions

So a number 

# 

```python
numbers_dict = {}

for i in range(1, 91):
    numbers_dict[str(i)] = 0

for index, df_row in enumerate(df_list):
    numbers = df_row[1:-2]
    for n in numbers:
        n_str = str(n)  
        if numbers_dict[n_str] == 0:
            numbers_dict[n_str] = index
        else:
            numbers_dict[n_str] = int(np.mean([index - numbers_dict[n_str], numbers_dict[n_str]]))

dict_items = list(numbers_dict.items())
dict_values = list(numbers_dict.values())
sorted_dict_items = sorted(dict_items,  key= lambda x : (x[1]))
print("Average extraction interval of a same number", round(np.mean(dict_values), 2))
print("Standard Deviation ", round(np.std(dict_values), 2))
print("Median ", round(np.median(dict_values), 2))
print("Max extraction interval ", max(dict_values))
print("Min extraction interval ", min(dict_values))
print("Most common numbers : ", sorted_dict_items[-5:])
print("Least common numbers : ", sorted_dict_items[:5])
```
