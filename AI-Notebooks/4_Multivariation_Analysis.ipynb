```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("iris.csv")
```

# 📊 Introduction to Multivariate Exploratory Analysis

In real-world datasets, there are many variables: we need to understand how they interact.

**Main goals:**
- Detect linear (or monotonic) relationships
- Identify redundant variables or collinearity
- Check for differences between groups

**This step is preparatory for:**
- Feature selection and feature engineering
- Choosing the most suitable ML algorithms

# 🔗 Correlation Matrix

The correlation matrix provides, at a glance, all pairwise correlations between quantitative variables.

Each value ranges from −1 to +1:
- **Close to +1**: strong positive linear relationship (when one increases, the other also increases)
- **Close to −1**: strong negative linear relationship (when one increases, the other decreases)
- **Close to 0**: no linear relationship

**Why it is useful:**
- **Redundancy**: if two variables have r ≈ 1, one of them adds little additional information
- **Multicollinearity**: groups of strongly correlated variables can complicate the use of linear models
- **Feature selection**: helps determine which features to include or transform



```python
corr_matrix = df.corr(numeric_only=True)
corr_matrix
```

```python
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', vmin=-1, vmax=1)
```

# 🌸 Pair Plot of the Iris Dataset — Exploratory Data Analysis

This pair plot visualizes the relationships between four quantitative features of the Iris dataset:  
- **Sepal length**  
- **Sepal width**  
- **Petal length**  
- **Petal width**  

Each point represents an individual flower, color-coded by species: *Setosa*, *Versicolor*, and *Virginica*.

## 🔍 How to read the plot
- **Diagonal plots** → Kernel Density Estimates (KDEs) showing the distribution of each feature for the three species.
- **Off-diagonal plots** → Scatterplots comparing each pair of features, allowing visual inspection of correlations and class separability.

## 📈 Insights from the plot
1. **Class separability**  
   - *Setosa* is clearly separated from the other two species in petal measurements.  
   - *Versicolor* and *Virginica* overlap more, suggesting classification between them may be harder.

2. **Correlations**  
   - Petal length and petal width show a strong positive correlation.  
   - Sepal width has weaker correlations with other features.

3. **Feature redundancy**  
   - Highly correlated features may be redundant in some models (e.g., linear regression).

4. **Feature selection guidance**  
   - Features with clear separation (e.g., petal length) are strong candidates for predictive modeling.  
   - Features with little separation may need transformation or could be less useful.
