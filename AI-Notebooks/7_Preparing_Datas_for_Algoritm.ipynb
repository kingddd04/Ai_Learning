# Preparing Datas For a ml-algoritm, scaling

Numerical variables can be expressed in very different units of measurement.  
**Example:** salary (in thousands of euros) vs age (in years). Salary values can reach up to 500,000, while age ranges from 0 to 100.

If you use an algorithm based on distances or numerical weights (such as linear regression, KNN, or clustering), the variable with larger values might dominate the determination of the result, even if it is not the most important one.

**Scaling** means bringing variables to comparable scales so that they all have the same initial "weight" in the analysis.

**Scaling** is a preprocessing technique that transforms feature values so they share a similar range or magnitude.  
It ensures that all features contribute equally to the model, preventing those with larger numeric ranges from dominating the learning process.

### 📌 Standardization (Z‑Score)

    scaled_variable (from dataset) = (variable - mean(variable)) / standard_deviation(variable)

This transformation centers the data around **0** and scales it so that the **standard deviation** is **1**.  
The resulting values (Z‑scores) represent how many standard deviations each observation is from the mean.

### 🔍 Why Use Standardization?
- **Equal footing for features**: Prevents large‑magnitude features from dominating the model.
- **Faster, more stable training**: Especially for gradient‑based algorithms.
- **Better performance**: Many models assume features are centered and scaled.
- **Works with outliers better than min–max scaling** (though still sensitive to extreme values).

### 📌 When It’s Important
- ✅ **Distance-based algorithms**: KNN, K‑Means, DBSCAN  
- ✅ **Gradient-based models**: Linear/Logistic Regression, SVM, Neural Networks  
- ✅ **Dimensionality reduction**: PCA, LDA  
- ❌ **Tree-based models**: Random Forest, XGBoost (scaling usually not needed)

### 📊 Interpreting Z‑Scores
- **0** → exactly at the mean  
- **+1** → 1 standard deviation above the mean  
- **−1** → 1 standard deviation below the mean  
- **>|3|** → often considered an outlier (in normally distributed data)

### ⚠️ Key Notes
- Z‑scores are **unbounded**: they can range from −∞ to +∞.
- In a normal distribution:
  - ~68% of values fall between −1 and +1
  - ~95% between −2 and +2
  - ~99.7% between −3 and +3

### 📌 Min–Max Scaling

    scaled_variable (from dataset) = (variable - min(variable)) / (max(variable) - min(variable))

Where:
- **min(variable)** = smallest value of the feature
- **max(variable)** = largest value of the feature

This transformation rescales the data so that all values fall within a fixed range — by default **[0, 1]**.  
It preserves the shape of the original distribution but compresses or stretches values to fit the target range.

### 🔍 Why Use Min–Max Scaling?
- **Equal footing for features**: Ensures all features contribute equally regardless of their original units or magnitude.
- **Preserves distribution shape**: Relative distances between values remain the same.
- **Essential for bounded input models**: Especially useful for algorithms or activation functions that expect inputs in a specific range (e.g., sigmoid, tanh).
- **Common in image processing**: Pixel values often normalized to [0, 1].


### 📌 When It’s Important
- ✅ **Distance-based algorithms**: KNN, K‑Means, DBSCAN  
- ✅ **Gradient-based models**: Neural Networks, Logistic Regression, SVM  
- ✅ **Image processing**: Scaling pixel intensities  
- ❌ **Tree-based models**: Random Forest, XGBoost (scaling usually not needed)

### 📊 Interpreting Min–Max Scaled Values
- **0** → corresponds to the minimum value in the feature  
- **1** → corresponds to the maximum value in the feature  
- Values between 0 and 1 are proportionally spaced between min and max

### ⚠️ Key Notes
- **Sensitive to outliers**: Extreme values can compress most data points into a narrow range.
- You can change the target range (e.g., `[-1, 1]`) using the `feature_range` parameter in `MinMaxScaler`.
- Always fit the scaler on the **training set only** and apply the same transformation to the test set to avoid **data leakage**.

```python
from sklearn.preprocessing import MinMaxScaler

minmax_matr = MinMaxScaler().fit_transform(df[["sepal.length","sepal.width","petal.length","petal.width"]])

minmax_df = pd.DataFrame(
    minmax_matr,
    columns=["sepal.length", "sepal.width", "petal.length", "petal.width"],
    index=df.index
)

print(minmax_matr[:5])
print(minmax_df.head(5))
```

### 📌 Robust Scaling

    scaled_variable (from dataset) = (variable - median(variable)) / IQR(variable)

Where:
- **median(variable)** = the middle value of the feature
- **IQR(variable)** = Interquartile Range = Q3 − Q1 (75th percentile − 25th percentile)

This transformation centers the data around **0** and scales it according to the spread of the middle 50% of the data, making it **less sensitive to outliers** than methods based on the mean and standard deviation.

### 🔍 Why Use Robust Scaling?
- **Resistant to outliers**: Median and IQR are not affected by extreme values.
- **Fair comparison**: Brings features to comparable scales without being skewed by anomalies.
- **Stable for skewed data**: Works well when distributions are not perfectly normal.
- **Better than Z‑score for heavy‑tailed data**: Avoids distortion from extreme points.

### 📌 When It’s Important
- ✅ Data with significant outliers  
- ✅ Skewed distributions  
- ✅ Distance‑based algorithms (KNN, K‑Means, DBSCAN)  
- ✅ Gradient‑based models (Linear/Logistic Regression, SVM, Neural Networks)  
- ❌ Tree‑based models (Random Forest, XGBoost) — scaling usually not needed

### 📊 Interpreting Robust‑Scaled Values
- **0** → exactly at the median  
- **+1** → 1 IQR above the median  
- **−1** → 1 IQR below the median  
- Values beyond ±1 indicate points outside the middle 50% of the data

### ⚠️ Key Notes
- Unlike Min–Max scaling, values are **not bounded** to a fixed range.
- More robust to outliers than Z‑score standardization, but still affected by extreme skew.

## 📊 Data Preparation: Encoding Categorical Variables

Real‑world datasets do not contain only numbers — they often include **categorical variables** (e.g., color, city, brand).  
Machine Learning algorithms cannot work directly with text or labels; they can only process numerical values.  

Therefore, we need a method to transform categories into numbers **without introducing incorrect or misleading meanings**.  

**Encoding** is the bridge that makes categorical variables readable and usable by ML models.

## 📊 Data Preparation: Ordinal Encoding

Some categorical variables have a **logical order** (e.g., risk level: low < medium < high).  
In this case, we can map the categories to numbers that are consistent with that order.  

This process is called **Ordinal Encoding**.

### 🔍 Why Use Ordinal Encoding?
- Preserves the **natural ranking** of categories.
- Allows algorithms to interpret the **relative order** between categories.
- More compact than one‑hot encoding for ordered variables.

### 📌 When to Use
- ✅ Ordered categories (e.g., education level, satisfaction rating, size: small < medium < large)  
- ❌ Unordered categories (e.g., colors, brands) — use one‑hot encoding instead

💡 **Tip:** Be careful — ordinal encoding introduces the assumption that the distance between categories is meaningful. Only use it when that assumption makes sense for your data.

```python
# Example DataFrame
df = pd.DataFrame({
    "livello": ["basso", "medio", "alto", "medio", "basso"]
})

# Mapping dictionary for ordinal encoding
mapping = {"basso": 1, "medio": 2, "alto": 3}

# Apply the mapping to the column
df["livello"] = df["livello"].map(mapping)

print(df)
```

## 📊 Data Preparation: One‑Hot Encoding

**One‑Hot Encoding** is a technique used to convert **categorical variables** into a numerical format that machine learning algorithms can understand — without introducing any false sense of order between categories.


### 🔍 Why Use One‑Hot Encoding?
- **Removes false ordinality**: Prevents the model from assuming a ranking between categories (e.g., "red" ≠ higher than "blue").
- **Model compatibility**: Many ML algorithms require purely numerical input.
- **Improved performance**: Allows models to capture relationships between categories more accurately.

### ⚙️ How It Works
For each category in a feature, One‑Hot Encoding creates a **new binary column**:
- **1** → the row belongs to that category
- **0** → the row does not belong to that category

Example:  
| Color   |  
|---------|  
| Red     |  
| Blue    |  
| Green   |  

After One‑Hot Encoding:  
| Color_Red | Color_Blue | Color_Green |  
|-----------|------------|-------------|  
| 1         | 0          | 0           |  
| 0         | 1          | 0           |  
| 0         | 0          | 1           |  


### 📌 When It’s Important
- ✅ **Unordered categorical variables** (e.g., colors, cities, brands)  
- ✅ Algorithms sensitive to numeric magnitude (e.g., Linear Regression, Logistic Regression, Neural Networks, KNN)  
- ❌ Not needed for ordered categories — use **Ordinal Encoding** instead

### ⚠️ Key Notes
- Can significantly increase the number of features if a variable has many unique categories (high cardinality).
- To avoid multicollinearity in linear models, you can drop one of the dummy columns (`drop_first=True` in pandas).
- Always fit the encoder on the **training set only** and apply it to the test set to avoid **data leakage**.
