## 📌 Pipeline in Machine Learning — What It Is & Why Use It

A **Pipeline** is a tool in machine learning that lets you chain together multiple processing steps  
(e.g., scaling, encoding, feature creation, model training) into **one single object**.

Instead of running each step manually, the pipeline ensures they happen:
- **In the correct order**
- **Consistently** during training, testing, and cross‑validation

### **Why We Need It?**
**Duplicated Code**  
   - Writing the same transformation code for training, testing, and each CV fold.
   - Inefficient and harder to maintain.

**Errors & Inconsistencies**  
   - Applying slightly different transformations to different datasets by mistake.
   - Example: scaling training data with one mean, but test data with another.

**Overfitting via Data Leakage**  
   - If you calculate transformation parameters (e.g., mean for scaling) using the **entire dataset** before splitting,  
     you let the model “peek” at the test data.
   - This contaminates evaluation and makes the model look better than it really is.

**How Pipelines Helps :**
- **Automation** → All steps run in sequence automatically.
- **Consistency** → Same transformations applied to training and test data.
- **Safety** → Prevents data leakage by fitting transformations only on the training fold during CV.
- **Cleaner Code** → One object handles everything — easier to read, debug, and reuse.

**So in the end :**
> Pipelines are not just a convenience — they are a **best practice** in ML workflows.  
> They protect against subtle but serious mistakes, keep your code clean, and ensure fair model evaluation.

## **-Pipeline Class in scikit-learn**
The `Pipeline` class in **scikit-learn** lets you **chain multiple processing steps** into a single object.  
Each step is defined as a **tuple**:
- A **name** (string)
- A **transformer** or **model** object
- Robustness → Reduces the chance of forgetting a preprocessing step or applying it incorrectly.
