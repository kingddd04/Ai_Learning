# Preventing Overfitting on complex models with Regularization

Complex models like polnomial Regression at high degrees may overfit and adapt too much to the datas to limit this; we can use some special linear regression models that use a technique called regolarization to perform better in certain fields we may choose from :

  - **Ridge** : Great if there are many features and they are all more or less relevant. It also works well when the features are strongly   
                correlated with each other, distributing the importance among them.
  - **Lasso** : The ideal choice if you think that only a subset of the features is truly relevant for the model.   
               It is useful for automatic feature selection.
  - **Elasticnet** : An excellent default choice. It combines the advantages of Ridge (handling multicollinearity)  
               and Lasso (feature selection). It is recommended  when there are many features and it is not known a priori which ones are relevant.

#### **Loading Sample Datas** :

```python
import pandas as pd

df = pd.read_csv('usa_houses.csv')
df = df[["price",'sqft_living', 'bedrooms', 'floors', 'view']]
df = df[(df['price'] < 2700000) & (df['price'] > 10000)]

features_df = df[['sqft_living', 'bedrooms', 'floors', 'view']]
target_df = df[["price"]]
```

### Adding polynomial features to Datas

```python
from sklearn.preprocessing import PolynomialFeatures
polynomial_features_df = PolynomialFeatures(degree=7).fit_transform(features_df)
```

### Standardizing Training set

```python
from sklearn.preprocessing import MinMaxScaler

standardized_polynomial_features_df = MinMaxScaler().fit_transform(polynomial_features_df)
```

### Training the Ridge Model 

📌 What is Ridge Regression?
Ridge Regression is a regularized version of linear regression that uses L2 regularization.
It’s designed to address two common problems in ordinary least squares (OLS) regression:
1. 	Overfitting — when the model fits noise in the training data and performs poorly on unseen data.
2. 	Multicollinearity — when predictor variables are highly correlated, making coefficient estimates unstable.

```python
from sklearn.linear_model import Ridge

ridge_model = Ridge(alpha=0.1) # alpha controls the strenght of regolarization
ridge_model.fit(standardized_polynomial_features_df, target_df)
#ridge_predicted_y = ridge_model.predict(polinomial_standardized_features_df)
ridge_r2_score = ridge_model.score(standardized_polynomial_features_df, target_df)
print("R2 Score (Train):",int(ridge_r2_score*100), "%")
```

# Training the Lasso Model

```python
from sklearn.linear_model import Lasso

lasso_model = Lasso(alpha=1, max_iter=1000)
lasso_model.fit(standardized_polynomial_features_df, target_df)
lasso_r2_score = lasso_model.score(standardized_polynomial_features_df, target_df)
print("R2 Score (Train):",int(lasso_r2_score*100), "%")
```

### Elasticnet

```python
from sklearn.linear_model import ElasticNet

elasticnet_model = ElasticNet(alpha=0.1, l1_ratio=0.5, max_iter=1000)
elasticnet_model.fit(standardized_polynomial_features_df, target_df)
elasticnet_r2_score = lasso_model = elasticnet_model.score(standardized_polynomial_features_df, target_df)
print("R2 Score (Train):",int(elasticnet_r2_score*100), "%")
```
