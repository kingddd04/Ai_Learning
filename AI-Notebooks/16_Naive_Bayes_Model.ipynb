### Why "Naive"?
"Naive" Because it assumes that **all features are conditionally independent** given the class label.  
This assumption is rarely true in real data, but it makes the model simple, fast, and surprisingly effective.

### 🧠 Naive Bayes Classifiers: Variants and Use Cases

Naive Bayes classifiers are a family of probabilistic models based on **Bayes' theorem** with the **naive assumption** that features are conditionally independent given the class.  
Different variants are tailored to specific data types and distributions.

## 📘 Multinomial Naive Bayes
- **Distribution**: Multinomial  
- **Feature type**: Discrete counts (e.g., word frequencies)  
- **How it works**: Models the probability of observing a certain count of features (like word counts in text). Each feature’s frequency is assumed to follow a multinomial distribution conditioned on the class.  
- **When to use**:  
  - Text classification with bag‑of‑words or TF‑IDF features  
  - Spam detection, sentiment analysis, topic classification  
- **Why it’s effective**: Captures how often words appear, not just whether they appear.

## 📗 Bernoulli Naive Bayes
- **Distribution**: Bernoulli (binary)  
- **Feature type**: Binary (0/1, present/absent)  
- **How it works**: Each feature is treated as a binary indicator (e.g., “word present in document” vs “not present”).  
- **When to use**:  
  - Text classification with binary word presence features  
  - Problems where frequency doesn’t matter, only presence/absence  
- **Why it’s effective**: Simpler than Multinomial NB when counts are not meaningful.

## 📙 Complement Naive Bayes
- **Distribution**: Modified Multinomial  
- **Feature type**: Discrete counts  
- **How it works**: Uses statistics from all classes *except* the target class to estimate probabilities. This adjustment helps correct biases.  
- **When to use**:  
  - Text classification with **imbalanced datasets**  
  - Cases where one class dominates the data  
- **Why it’s effective**: More robust than Multinomial NB when classes are skewed.

## 📒 Categorical Naive Bayes
- **Distribution**: Categorical  
- **Feature type**: Categorical (non‑numeric labels)  
- **How it works**: Each feature is modeled as a categorical variable with a finite set of possible values.  
- **When to use**:  
  - Datasets with categorical features (e.g., color = red/green/blue, brand = A/B/C)  
- **Why it’s effective**: Handles non‑numeric data directly without encoding into counts or binaries.


## 📕 Gaussian Naive Bayes
- **Distribution**: Gaussian (Normal)  
- **Feature type**: Continuous  
- **How it works**: Assumes each feature follows a normal distribution within each class. Learns the mean and variance per class per feature.  
- **When to use**:  
  - Datasets with continuous, roughly bell‑shaped features  
  - Classic example: Iris dataset (petal length, petal width, etc.)  
- **Why it’s effective**: Simple, fast, and often surprisingly accurate even when independence assumption is violated.


## 📊 Summary Table: Naive Bayes Variants

| Variant               | Feature Type     | Distribution Assumption | Typical Use Case                     |
|-----------------------|------------------|--------------------------|--------------------------------------|
| **Multinomial NB**    | Discrete counts  | Multinomial              | Text classification (word counts)    |
| **Bernoulli NB**      | Binary           | Bernoulli                | Binary text features (word presence) |
| **Complement NB**     | Discrete counts  | Complement Multinomial   | Imbalanced text classification       |
| **Categorical NB**    | Categorical      | Categorical              | Categorical feature classification   |
| **Gaussian NB**       | Continuous       | Gaussian (Normal)        | Continuous feature classification    |



### 📌 Summary
- **Gaussian** → continuous features (assumed normal).  
- **Multinomial** → discrete counts/frequencies.  
- **Bernoulli** → binary features.  
- **Complement** → imbalanced text datasets.  
- **Categorical** → categorical features.  

## ⚖️ Pros and Cons of Naive Bayes Classifiers

Naive Bayes classifiers are simple yet powerful probabilistic models.  
They are often used as **baseline models** in classification tasks, especially in text mining and natural language processing.  
Here are their main strengths and weaknesses:

### ✅ Advantages
- **Fast training and prediction** → Only requires calculating means, variances, and priors.  
- **Works well with high-dimensional data** → Common in text classification (bag-of-words, TF-IDF).  
- **Requires relatively little training data** → Performs well even with small datasets.  
- **Simple and interpretable** → Easy to understand and explain.  
- **Effective in text classification and spam detection** → A classic choice for NLP tasks.  
- **Handles missing data gracefully** (in some implementations) → Can ignore features that are missing during prediction.  

### ❌ Disadvantages
- **Strong independence assumption** → Rarely true in real-world data, since features are often correlated.  
- **Performance degrades with correlated features** → Violates the “naive” assumption.  
- **Not suitable for complex decision boundaries** → Linear decision surfaces may underfit complex patterns.  
- **Sensitive to feature distribution assumptions** → e.g., Gaussian NB assumes normality, which may not hold.  
- **May underperform compared to more flexible models** → Random Forests, SVMs, or Neural Networks often achieve higher accuracy.  

### Implementation of Gaussian Naive Bayes

**Importing the dataset**

```python
import pandas as pd

df = pd.read_csv('iris.csv')

x = df.drop("variety" , axis=1)
y = df['variety']
```

### Model Training 

```python
from sklearn.naive_bayes import GaussianNB

gaussian_naive_bayes_model = GaussianNB()

gaussian_naive_bayes_model.fit(x,y)
```

### Getting a Prediction

```python
from sklearn.metrics import classification_report

y_pred = gaussian_naive_bayes_model.predict(x)

print("GaussianNB Results:\n", classification_report(y, y_pred, target_names=df['variety'].unique()))
```
