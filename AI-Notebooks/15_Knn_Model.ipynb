### KNN Model (k_nearest neighbours)

The K-Nearest Neighbors (KNN) is a non-parametric, supervised learning algorithm based on instances.  
It does not build an explicit model: to classify a new point, it looks at the K nearest examples    
in the training dataset and decides the class based on them. This type of model does not require training for obious reasons.

The distance may be the mikovoski distance, manatthan distance and the Cosine distance 

Knn Is a lazy learning algoritm in a sense that it does not process datas until a prediction is required.
Knn Can be used for both classification and regression 

**KNN for Regression :** Predict the target value as the **average (or weighted average)**  
of the `k` nearest neighbors.

###  KNN with Multiple Features

In KNN, each sample is represented as a **vector in feature space**.  
For example, a house might be described by:

- `square_feet`
- `floors`
- `bedrooms`

This makes each house a point in a **3D space**.  
KNN computes the **distance** (usually Euclidean) between the query point and all training points:

- **Classification** → the class is decided by the **majority vote** of the `k` nearest neighbors.  
- **Regression** → the target value is the **average (or weighted average)** of the `k` nearest neighbors.  

Note that calssification and regression have different evaluation metrics. Classification has class report and regression have r2, mse, rmse  
⚠️ Since distances depend on feature scales, **normalization/standardization** is essential (e.g., square feet would otherwise dominate over floors/bedrooms).

### ✅ Advantages
- **Simple & Intuitive** → Easy to explain and implement, no training phase required.  
- **Flexible** → Works for both classification and regression tasks.  
- **Non-parametric** → No assumptions about data distribution or linearity.  
- **Naturally Multiclass** → Handles multiple classes without modification.  
- **Local Adaptability** → Captures complex decision boundaries when `k` is small.  
- **Interpretable** → You can inspect which neighbors influenced a prediction.  

### ❌ Disadvantages
- **Prediction is Slow** → Must compute distances to all training points (costly for large datasets).  
- **Sensitive to Feature Scaling** → Requires normalization/standardization.  
- **Curse of Dimensionality** → Performance degrades in high-dimensional spaces.  
- **Noisy with Low k** → Small `k` can lead to overfitting and instability.  
- **Majority Class Bias** → Struggles with imbalanced datasets (minority class gets outvoted).  
- **Memory-Intensive** → Needs to store the entire training set.  
- **Sensible to OutLayers** an extreme example may mess the prediction

Consequences of small and high k depends on the dataset size and class distribution so it's important to have a general idea of the dataset before setting the K value.
We should consider using cross-validation and consider distance-weighted voting or resampling to mitigate imbalance effect.

### ⚖️ Best Use Cases
- Small to medium datasets  
- Low-dimensional feature spaces  
- Well-preprocessed and scaled data  
- When interpretability and simplicity are more important than speed  

### Inplementing a Classification knn (Sklearn)

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split

df = pd.read_csv("iris.csv")

x_classification = df.drop(columns=["variety"])
y_classification = df["variety"]

x_classification_train, x_classification_test, y_classification_train, y_classification_test = train_test_split(
    x_classification, y_classification, test_size=0.3, random_state=42 ,stratify=y_classification
)


df["variety"].value_counts()
```

### Scaling Datas 

We must create a scaler on train set and evaluate it on test set

```python

from sklearn.preprocessing import StandardScaler
"""
x_features_classification_train_sc = StandardScaler().fit_transform(x_classification_train)
x_features_classification_test_sc = StandardScaler().fit_transform(x_classification_test)
"""
sc = StandardScaler()
x_features_classification_train_sc = sc.fit_transform(x_classification_train)
x_features_classification_test_sc = sc.transform(x_classification_test)
```

### Setting up the model and "training"

```python
from sklearn.neighbors import KNeighborsClassifier

knn_classifier = KNeighborsClassifier(
    n_neighbors=5,        # Number of neighbors (k). Here: look at 3 closest points.
    weights="distance",   # Weighting scheme: closer neighbors have more influence.
    metric="euclidean",   # Distance metric: standard Euclidean (L2 norm).
    algorithm="auto",     # Search algorithm: scikit-learn chooses best (kd_tree, ball_tree, brute).
)

knn_classifier.fit(x_features_classification_train_sc, y_classification_train)
```

### Getting a Prediction and Evaluation

```python
from sklearn.metrics import confusion_matrix, classification_report

y_classification_pred = knn_classifier.predict(x_features_classification_test_sc)

print(classification_report(y_classification_test, y_classification_pred, digits=3))
```
